```python
songs = ["Kesariya", "Believer", "Shape of You", "Blinding Lights", "Excuses"]

with open("my_playlist.txt", "w") as file:
    for song in songs:
        file.write(song + "\n")

print("my_playlist.txt created successfully.")
```

```python
with open("my_playlist.txt", "r") as file:
    for song in file:
        print(song.strip().upper())
```

```python
import csv

# Sample IPL match data
matches = [
    ["Match", "Team1", "Team2", "Winner"],
    ["Match 1", "CSK", "MI", "CSK"],
    ["Match 2", "RCB", "KKR", "KKR"],
    ["Match 3", "GT", "RR", "GT"]
]

with open("ipl_match_scores.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(matches)

with open("ipl_match_scores.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print("Winning Team:", row["Winner"])
```

```python
import json

user_profile = {
    "username": "tej_user",
    "followers": 1500,
    "bio": "Data Analytics Student"
}

with open("user_profile.json", "w") as file:
    json.dump(user_profile, file, indent=4)

with open("user_profile.json", "r") as file:
    profile = json.load(file)

print("Username:", profile["username"])
print("Followers:", profile["followers"])
```
