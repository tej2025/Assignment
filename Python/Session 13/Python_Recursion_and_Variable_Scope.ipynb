```python
def reverse_string(s):
    if len(s) == 0:
        return s
    return reverse_string(s[1:]) + s[0]

print(reverse_string("hello"))
```

```python
def sum_playlist_durations(durations):
    if len(durations) == 0:
        return 0
    return durations[0] + sum_playlist_durations(durations[1:])

durations = [210, 185, 240, 200]

print("Total duration:", sum_playlist_durations(durations), "seconds")
```

```python
count = 10

def update_count():
    count = 5
    print("Inside:", count)

update_count()
print("Outside:", count)

# 'count' inside update_count() is a local variable.
# The 'count' outside the function is a global variable.
# Output:
# Inside: 5
# Outside: 10
```

```python
def count_likes(posts):
    total = 0

    for post in posts:
        total += post.get("likes", 0)

        replies = post.get("replies", [])
        total += count_likes(replies)

    return total

posts = [
    {
        "likes": 100,
        "replies": [
            {
                "likes": 20,
                "replies": [
                    {"likes": 5, "replies": []}
                ]
            },
            {"likes": 10, "replies": []}
        ]
    },
    {
        "likes": 50,
        "replies": [
            {"likes": 15, "replies": []}
        ]
    }
]

print("Total likes:", count_likes(posts))
```

```python
app_status = "Offline"

def show_status():
    user_status = "Online"
    print("During function - user_status:", user_status)
    print("During function - app_status:", app_status)

print("Before function - app_status:", app_status)

show_status()

print("After function - app_status:", app_status)

# user_status is a local variable and exists only during the function call.
# app_status is a global variable and remains available after the function call.
```
