```python
movies = ["Dune", "Avatar", "Inception", "Interstellar", "Oppenheimer"]

movie_iterator = iter(movies)

while True:
    try:
        movie = next(movie_iterator)
        print(movie)
    except StopIteration:
        break
```

```python
songs = ["Kesariya", "Believer", "Shape of You", "Blinding Lights", "Excuses", "Perfect"]

for position, song in enumerate(songs, start=1):
    print(f"{position}. {song}")
```

```python
food_items = ["Pizza", "Burger", "Pasta", "Sandwich"]
prices = [250, 180, 220, 150]

for food, price in zip(food_items, prices):
    print(f"{food} - ₹{price}")
```

```python
def insta_posts_generator(posts):
    for post in posts:
        yield post

posts = [
    "Weekend vibes!",
    "New day, new goals!",
    "Exploring new places!"
]

post_generator = insta_posts_generator(posts)

while True:
    try:
        print(next(post_generator))
    except StopIteration:
        break
```

```python
def cashback_generator(transactions):
    for amount in transactions:
        yield amount * 0.05

transactions = [500, 1000, 750, 2000]

for cashback in cashback_generator(transactions):
    print("Cashback:", cashback)
```
