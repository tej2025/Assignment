```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero"

print(safe_divide(10, 2))
print(safe_divide(10, 0))
```

```python
try:
    number_of_reviews = int(input("Enter number of reviews: "))
    total_stars = int(input("Enter total stars: "))

    average_rating = total_stars / number_of_reviews
    print("Average rating:", average_rating)

except ValueError:
    print("Error: Please enter numeric values.")
except ZeroDivisionError:
    print("Error: Number of reviews cannot be zero.")
```

```python
class InvalidDurationError(Exception):
    pass

def get_playlist_duration(songs):
    if any(duration < 0 for duration in songs):
        raise InvalidDurationError("Song duration cannot be negative.")

    total_seconds = sum(songs)
    return total_seconds / 60

try:
    durations = [210, 180, 240, 200]
    print("Total duration:", get_playlist_duration(durations), "minutes")
except InvalidDurationError as e:
    print("Error:", e)
```

```python
try:
    item_price = float(input("Enter item price: "))
    quantity = int(input("Enter quantity: "))

    total_price = item_price * quantity

except ValueError:
    print("Error: Please enter valid numeric values.")

else:
    print("Total price:", total_price)

finally:
    print("Thank you for shopping!")
```
