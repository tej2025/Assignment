```python
def remove_last_item(order_list):
    removed_item = order_list.pop()
    return removed_item

order_list = ['pizza', 'burger', 'cold drink']

removed_item = remove_last_item(order_list)

print("Removed item:", removed_item)
print("Updated order:", order_list)
```
