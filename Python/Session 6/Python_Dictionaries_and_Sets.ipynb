```python
playlist_prices = {
    "Workout Mix": 99,
    "Chill Vibes": 79,
    "Party Hits": 89,
    "Focus Music": 69,
    "Travel Songs": 59
}

print(playlist_prices)
```

```python
def update_playlist_price(playlist, new_price):
    playlist_prices[playlist] = new_price

update_playlist_price("Party Hits", 109)

print(playlist_prices)
```

```python
del playlist_prices["Travel Songs"]

print(playlist_prices)
```

```python
set1 = {"Dominos", "McDonalds", "Subway", "Burger King"}
set2 = {"Subway", "Burger King", "KFC", "Pizza Hut"}

union_set = set1.union(set2)
intersection_set = set1.intersection(set2)

print("Union:", union_set)
print("Intersection:", intersection_set)
```
