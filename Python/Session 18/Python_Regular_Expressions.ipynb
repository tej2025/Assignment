```python
import re

text = """
Contact me at 9876543210 or 8123456789.
Random numbers: 1234567890, 500, 98765.
You can also call 7654321098.
"""

phone_numbers = re.findall(r"\b[789]\d{9}\b", text)

print(phone_numbers)
```

```python
import re

def contains_valid_date(text):
    pattern = r"\b\d{2}/\d{2}/\d{4}\b"
    return re.search(pattern, text) is not None

print(contains_valid_date("Order placed on 25/06/2024"))
print(contains_valid_date("Order placed yesterday"))
```

```python
import re

review_text = """
Great food! Contact support@example.com or write to
customer.help@gmail.com. Another email is user123@yahoo.com.
"""

emails = re.findall(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}", review_text)

print(emails)
```

```python
import re

text = "My Paytm number is 9876543210 and alternate number is 8123456789."

masked_text = re.sub(r"\b\d{6}(?=\d{4}\b)", "******", text)

print(masked_text)
```

```python
import re

# AI-generated regex pattern for the given order ID format:
# OD followed by exactly 18 digits
pattern = r"\bOD\d{18}\b"

order_strings = [
    "Your order ID is OD123456789012345000",
    "Order ID: OD987654321098765432",
    "Invalid order: AB123456789012345000"
]

for order in order_strings:
    match = re.search(pattern, order)

    if match:
        print("Valid order ID:", match.group())
    else:
        print("No valid order ID found.")
```
