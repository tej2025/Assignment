```python
order_amounts = [120, 250, 90, 310, 150]

total_order_value = 0

for amount in order_amounts:
    total_order_value += amount

print("Total order value:", total_order_value)
```

```python
scores = [45, 78, 102, 34, 67, 89]

i = 0

while i < len(scores) and scores[i] <= 100:
    print(scores[i])
    i += 1
```

```python
item_prices = [299, 499, 199, 999, 149]

total = 0

for price in item_prices:
    if price < 200:
        continue
    total += price

print("Total of remaining items:", total)
```

```python
songs = ['Kesariya', 'Believer', 'Shape of You', 'Blinding Lights', 'Excuses']

for position, song in enumerate(songs, start=1):
    print(position, song)
```

```python
followers = [120, 1500, 23000, 800, 45000]

for count in followers:
    if count < 1000:
        category = "Micro"
    elif count <= 10000:
        category = "Influencer"
    else:
        category = "Celebrity"

    print(count, category)
```
