```python
import pandas as pd

users = pd.DataFrame({
    'user_id': [1, 2, 3, 4, 5],
    'username': ['Tej', 'Rahul', 'Priya', 'Amit', 'Neha'],
    'city': ['Ahmedabad', 'Mumbai', 'Pune', 'Delhi', 'Surat']
})

orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105, 106, 107],
    'user_id': [1, 2, 1, 3, 4, 2, 3],
    'order_amount': [500, 750, 300, 900, 450, 1200, 650]
})

inner_join = pd.merge(users, orders, on='user_id', how='inner')

print(inner_join)
```

```python
left_join = pd.merge(users, orders, on='user_id', how='left')

print(left_join)
```

```python
restaurants = pd.DataFrame({
    'restaurant_id': [101, 102, 103, 104],
    'name': ['Food Hub', 'Spice House', 'Pizza Point', 'Biryani House'],
    'city': ['Ahmedabad', 'Mumbai', 'Pune', 'Surat']
})

zomato_reviews = pd.DataFrame({
    'review_id': [1, 2, 3, 4, 5],
    'restaurant_id': [101, 102, 105, 101, 103],
    'rating': [5, 4, 3, 4, 5],
    'reviewer_city': ['Ahmedabad', 'Mumbai', 'Delhi', 'Ahmedabad', 'Pune']
})

outer_join = pd.merge(restaurants, zomato_reviews, on='restaurant_id', how='outer')

print(outer_join)
```

```python
playlists = pd.DataFrame({
    'playlist_id': [201, 202, 203, 204, 205],
    'user_id': [1, 2, 3, 4, 5],
    'city': ['Ahmedabad', 'Mumbai', 'Pune', 'Delhi', 'Surat'],
    'genre': ['Pop', 'Rock', 'Hip-Hop', 'Bollywood', 'EDM']
})

user_profiles = pd.DataFrame({
    'user_id': [1, 2, 3, 4, 5],
    'username': ['Tej', 'Rahul', 'Priya', 'Amit', 'Neha'],
    'city': ['Ahmedabad', 'Mumbai', 'Delhi', 'Delhi', 'Vadodara']
})

matched_playlists = pd.merge(
    playlists,
    user_profiles,
    on=['user_id', 'city'],
    how='inner'
)

print(matched_playlists)
```

```python
flipkart_products = pd.DataFrame({
    'product_id': [101, 102, 103, 104],
    'name': ['Laptop', 'Headphones', 'Smartphone', 'Keyboard'],
    'category': ['Electronics', 'Audio', 'Electronics', 'Accessories']
})

product_ratings = pd.DataFrame({
    'rating_id': [1, 2, 3, 4, 5],
    'product_id': [101, 102, 105, 103, 105],
    'rating': [5, 4, 3, 5, 2]
})

# Changed from the default inner join to a right join
right_join = pd.merge(
    flipkart_products,
    product_ratings,
    on='product_id',
    how='right'
)

print(right_join)
```
