```python
import pandas as pd

instagram_users = pd.DataFrame({
    'username': ['CoolDude', 'foodieQueen', 'TravelLover', 'Tech_Guru', 'FitnessFan']
})

instagram_users['username'] = instagram_users['username'].str.lower()

print(instagram_users)
```

```python
songs = pd.Series([
    'Perfect Remix',
    'Believer Remix',
    'Shape of You',
    'Faded Remix Version',
    'Levitating Remix'
])

cleaned_songs = songs.str.replace('Remix', '', regex=False).str.replace('  ', ' ', regex=False).str.strip()

print(cleaned_songs)
```

```python
zomato_reviews = pd.DataFrame({
    'review': [
        'Food was great and delivery was fast',
        'The Delivery was very late',
        'Amazing taste and good packaging',
        'Quick delivery and fresh food',
        'Restaurant service was excellent'
    ]
})

delivery_reviews = zomato_reviews[
    zomato_reviews['review'].str.contains('delivery', case=False, na=False)
]

print(delivery_reviews)
```

```python
flipkart_categories = pd.Series([
    'Electronics/Mobiles/Smartphones',
    'Home/Kitchen/Appliances',
    'Fashion/Men/Shirts',
    'Beauty/Makeup/Lipstick',
    'Sports/Fitness/Equipment'
])

category_df = pd.DataFrame({'CategoryPath': flipkart_categories})
category_df['MainCategory'] = category_df['CategoryPath'].str.split('/').str[0]

print(category_df)
```

```python
youtube_df = pd.DataFrame({
    'title': [
        'How to Learn Python',
        'Best Excel Tips',
        'MySQL Tutorial for Beginners',
        'Power BI Dashboard Guide',
        'Data Analysis Project'
    ]
})

youtube_df['title'] = youtube_df['title'].str.upper().str.replace(' ', '_', regex=False)

print(youtube_df)
```
