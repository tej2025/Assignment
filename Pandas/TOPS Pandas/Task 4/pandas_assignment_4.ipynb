```python
import pandas as pd

movies = pd.DataFrame({
    'title': ['3 Idiots', 'Dangal', 'Sholay', 'Lagaan', 'Zindagi Na Milegi Dobara', 'Taare Zameen Par'],
    'rating': [8.4, 8.3, 8.1, 8.1, 8.2, 8.3],
    'release_year': [2009, 2016, 1975, 2001, 2011, 2007]
})

sorted_movies = movies.sort_values('rating', ascending=False)

print(sorted_movies)
```

```python
ipl_teams = pd.DataFrame({
    'team': ['Mumbai Indians', 'CSK', 'RCB', 'Gujarat Titans', 'Rajasthan Royals'],
    'points': [16, 15, 14, 13, 12],
    'net_run_rate': [0.450, 0.320, 0.180, 0.120, 0.050]
})

sorted_teams = ipl_teams.sort_index(ascending=False)

print(sorted_teams)
```

```python
youtube_videos = pd.DataFrame({
    'video_title': ['Music Video', 'Cricket Highlights', 'Comedy Show', 'Tech Review'],
    'views': [2500000, 1800000, 1200000, 950000],
    'likes': [180000, 125000, 95000, 70000]
}, index=[5, 8, 12, 15])

youtube_videos = youtube_videos.reset_index(drop=True)

print(youtube_videos)
```

```python
zomato = pd.DataFrame({
    'restaurant': ['Spice Hub', 'Pizza Point', 'Burger Cafe', 'Biryani House', 'Pasta Street'],
    'rating': [4.2, 4.0, 3.8, 4.5, 4.1]
})

zomato = zomato.set_index('restaurant')

new_order = ['Biryani House', 'Spice Hub', 'Pizza Point', 'Burger Cafe', 'Pasta Street']

zomato = zomato.reindex(new_order)

print(zomato)
```

```python
flipkart = pd.DataFrame({
    'product_name': ['Laptop', 'Smartphone', 'Headphones', 'Smartwatch', 'Tablet'],
    'price': [55000, 25000, 1800, 3200, 18000],
    'discount': [15, 25, 40, 30, 20]
})

flipkart = flipkart.sort_values('discount', ascending=False)
flipkart = flipkart.reset_index(drop=True)

print(flipkart)
```
