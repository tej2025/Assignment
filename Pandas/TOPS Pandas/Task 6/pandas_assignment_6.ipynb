```python
import pandas as pd

spotify_df = pd.read_csv('spotify_song_streams.csv')

print('Before removing duplicates:')
print(spotify_df)

spotify_clean = spotify_df.drop_duplicates(subset='song', keep='first')

print('\nAfter removing duplicate songs:')
print(spotify_clean)
```

```python
flipkart_reviews = pd.DataFrame({
    'product_id': [101, 102, 101, 103, 102, 104, 101],
    'user_id': [1, 2, 1, 3, 2, 4, 5],
    'rating': [5, 4, 5, 3, 4, 5, 2],
    'review_text': ['Excellent', 'Good', 'Excellent', 'Average', 'Good', 'Excellent', 'Bad']
})

duplicate_count_before = flipkart_reviews.duplicated(subset=['product_id', 'user_id']).sum()
print('Number of duplicate reviews before:', duplicate_count_before)

flipkart_clean = flipkart_reviews.drop_duplicates(subset=['product_id', 'user_id'], keep='first')

duplicate_count_after = flipkart_clean.duplicated(subset=['product_id', 'user_id']).sum()
print('Number of duplicate reviews after:', duplicate_count_after)
```

```python
zomato_orders = pd.DataFrame({
    'order_id': [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008],
    'order_amount': [250, 320, 280, 400, 350, 300, 2500, 290]
})

Q1 = zomato_orders['order_amount'].quantile(0.25)
Q3 = zomato_orders['order_amount'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = zomato_orders[
    (zomato_orders['order_amount'] < lower_bound) |
    (zomato_orders['order_amount'] > upper_bound)
]

print('Q1:', Q1)
print('Q3:', Q3)
print('IQR:', IQR)
print('Lower Bound:', lower_bound)
print('Upper Bound:', upper_bound)
print('\nOutlier Order IDs:')
print(outliers['order_id'].to_list())
```

```python
from scipy.stats import zscore

ipl_tickets = pd.DataFrame({
    'date': ['2026-04-01', '2026-04-02', '2026-04-03', '2026-04-04', '2026-04-05',
             '2026-04-06', '2026-04-07', '2026-04-08'],
    'tickets_sold': [1200, 1350, 1280, 1400, 1320, 1250, 5000, 1300]
})

ipl_tickets['z_score'] = zscore(ipl_tickets['tickets_sold'])

outlier_dates = ipl_tickets[ipl_tickets['z_score'].abs() > 2]

print('Dates with absolute Z-score greater than 2:')
print(outlier_dates[['date', 'tickets_sold', 'z_score']])
```

```python
from scipy.stats import zscore

def find_outlier_transactions(df):
    df = df.copy()
    df['z_score'] = zscore(df['transaction_amount'])
    return df[df['z_score'].abs() > 2]

paytm_wallet = pd.DataFrame({
    'transaction_id': [1, 2, 3, 4, 5, 6, 7, 8],
    'transaction_amount': [200, 250, 180, 220, 210, 230, 2400, 190]
})

outlier_transactions = find_outlier_transactions(paytm_wallet)

print('Outlier transactions:')
print(outlier_transactions)
```
