```python
import pandas as pd

sales_df = pd.read_csv('flipkart_product_sales.csv')

sales_df['TotalValue'] = sales_df['Price'] * sales_df['Qty']

print(sales_df)
```

```python
food_df = pd.DataFrame({
    'Item': ['Pizza', 'Burger', 'Pasta', 'Sandwich', 'Biryani'],
    'Price': [300, 200, 250, 150, 280],
    'Qty': [2, 3, 1, 2, 1]
})

food_df['DiscountedPrice'] = food_df['Price'].apply(lambda x: x * 0.90)

print(food_df)
```

```python
instagram_df = pd.DataFrame({
    'PostID': [101, 102, 103, 104],
    'Likes': [1500, 2500000, 850, 1200000],
    'Comments': [120, 15000, 75, 2300]
})

def format_number(x):
    if x >= 1000000:
        return f'{x / 1000000:.1f}M'
    elif x >= 1000:
        return f'{x / 1000:.1f}K'
    else:
        return str(x)

instagram_df[['Likes', 'Comments']] = instagram_df[['Likes', 'Comments']].applymap(format_number)

print(instagram_df)
```

```python
zomato_df = pd.DataFrame({
    'Restaurant': ['Pizza Hub', 'Burger Point', 'Biryani House', 'Pasta Corner'],
    'Price': [300, 200, 280, 250],
    'Qty': [2, 3, 1, 2],
    'DeliveryCharge': [40, 30, 50, 35]
})

zomato_df['FinalAmount'] = zomato_df.apply(
    lambda row: (row['Price'] * row['Qty']) + row['DeliveryCharge'], axis=1
)

print(zomato_df)
```

```python
myntra_df = pd.DataFrame({
    'OrderID': [1, 2, 3, 4],
    'Product': ['T-Shirt', 'Jeans', 'Shoes', 'Jacket'],
    'Price': [800, 1500, 2500, 2200],
    'Qty': [2, 1, 1, 2]
})

myntra_df['GST'] = myntra_df['Price'] * myntra_df['Qty'] * 0.05

print(myntra_df)
```

### What I learned

1. I learned that pandas can create a calculated column directly from existing columns using simple arithmetic.
2. I also learned that `apply()` and lambda functions are useful for applying a calculation row-by-row or value-by-value.
