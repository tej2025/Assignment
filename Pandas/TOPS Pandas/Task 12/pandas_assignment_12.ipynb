```python
import pandas as pd

ipl_df = pd.read_csv('ipl_team_runs.csv')

team_runs = ipl_df.groupby('team', as_index=False)['runs'].sum()
team_runs = team_runs.rename(columns={'runs': 'total_runs'})

print(team_runs)
```

```python
zomato_orders = pd.DataFrame({
    'city': ['Ahmedabad', 'Ahmedabad', 'Mumbai', 'Mumbai', 'Pune', 'Pune', 'Ahmedabad'],
    'restaurant': ['Food Hub', 'Food Hub', 'Pizza Point', 'Pizza Point', 'Food Hub', 'Food Hub', 'Spice House'],
    'order_amount': [400, 600, 500, 700, 450, 550, 350]
})

avg_order = zomato_orders.groupby(['city', 'restaurant']).agg(
    average_order_amount=('order_amount', 'mean')
).reset_index()

print(avg_order)
```

```python
flipkart_sales = pd.DataFrame({
    'category': ['Electronics', 'Electronics', 'Fashion', 'Fashion', 'Electronics', 'Fashion'],
    'region': ['North', 'South', 'North', 'South', 'North', 'South'],
    'sales': [50000, 65000, 30000, 28000, 45000, 35000],
    'units_sold': [50, 65, 30, 28, 45, 35]
})

sales_summary = flipkart_sales.groupby(['category', 'region']).agg(
    total_sales=('sales', 'sum'),
    average_units_sold=('units_sold', 'mean')
).reset_index()

print(sales_summary)
```

```python
revenue_df = pd.DataFrame({
    'region': ['North', 'North', 'South', 'South', 'West', 'West', 'East', 'East'],
    'revenue': [50000, 65000, 45000, 70000, 55000, 60000, 40000, 48000]
})

revenue_summary = revenue_df.groupby('region').agg({
    'revenue': ['sum', 'max', 'min']
})

print(revenue_summary)
```

```python
spotify_df = pd.DataFrame({
    'artist': ['Artist A', 'Artist B', 'Artist C', 'Artist D', 'Artist E', 'Artist F'],
    'genre': ['Pop', 'Rock', 'Pop', 'Hip-Hop', 'Rock', 'Pop'],
    'streams': [5000000, 3500000, 4200000, 2800000, 3000000, 6000000]
})

genre_summary = spotify_df.groupby('genre').agg(
    total_streams=('streams', 'sum'),
    average_streams=('streams', 'mean')
).sort_values('total_streams', ascending=False)

print(genre_summary)
```
