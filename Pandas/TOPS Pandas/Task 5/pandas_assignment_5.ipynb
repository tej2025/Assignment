```python
import pandas as pd

ipl_players = pd.read_csv('ipl_players.csv')

missing_age = ipl_players[ipl_players['age'].isna()]

print(missing_age)
```

```python
food_orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105],
    'customer_name': ['Rahul', 'Priya', 'Amit', 'Neha', 'Karan'],
    'delivery_rating': [4.5, None, 4.0, 3.5, None]
})

cleaned_orders = food_orders.dropna(subset=['delivery_rating'])

print(cleaned_orders)
```

```python
product_reviews = pd.DataFrame({
    'product_name': ['Laptop', 'Phone', 'Headphones', 'Watch', 'Tablet'],
    'rating': [4.5, 4.0, None, 3.5, 4.0]
})

mean_rating = product_reviews['rating'].mean()

product_reviews['rating'] = product_reviews['rating'].fillna(mean_rating)

print(product_reviews)
```

```python
spotify = pd.read_csv('spotify_playlist.csv')

median_duration = spotify['duration'].median()

spotify['duration'] = spotify['duration'].fillna(median_duration)

spotify.to_csv('cleaned_spotify_playlist.csv', index=False)

print(spotify)
```

```python
instagram = pd.read_csv('instagram_users.csv')

instagram['followers'] = instagram['followers'].fillna(0)

print(instagram)
```
