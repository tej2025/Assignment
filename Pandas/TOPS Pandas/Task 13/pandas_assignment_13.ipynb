```python
import pandas as pd

ipl_df = pd.read_csv('IPL_2023_Batsmen.csv')

print(ipl_df.head(10))
```

```python
print('Mean:', ipl_df['runs'].mean())
print('Median:', ipl_df['runs'].median())
print('Mode:', ipl_df['runs'].mode().tolist())
print('Minimum:', ipl_df['runs'].min())
print('Maximum:', ipl_df['runs'].max())
```

```python
zomato = pd.DataFrame({
    'name': ['Restaurant A', 'Restaurant B', 'Restaurant C', 'Restaurant D', 'Restaurant E'],
    'rating': [4.5, 4.2, 4.7, 3.9, 4.4],
    'votes': [1200, 850, 1500, 500, 1100],
    'avg_cost': [600, 450, 800, 350, 550]
})

correlation = zomato['rating'].corr(zomato['votes'])

print(zomato)
print('\nCorrelation between rating and votes:', correlation)
```

```python
summary = ipl_df.describe()
print(summary)

print('\nInsight: The maximum runs value is much higher than the minimum, showing a large difference in batting output among players.')
```

```python
runs_strike_rate_corr = ipl_df['runs'].corr(ipl_df['strike_rate'])

print('Correlation between runs and strike rate:', runs_strike_rate_corr)
```
