```python
import pandas as pd
import numpy as np

ipl_df = pd.read_csv('ipl_matches_assignment14.csv')

print(ipl_df.head())
```

```python
ipl_df = ipl_df.dropna(subset=['winner']).reset_index(drop=True)

print(ipl_df)
```

```python
conditions = [
    ipl_df['win_by_runs'] > 0,
    ipl_df['win_by_wickets'] > 0
]

choices = ['Runs', 'Wickets']

ipl_df['match_margin_type'] = np.select(
    conditions,
    choices,
    default='Tie/No Result'
)

print(ipl_df[['team1', 'team2', 'winner', 'win_by_runs', 'win_by_wickets', 'match_margin_type']])
```

```python
team_cities = pd.DataFrame({
    'team': [
        'Mumbai Indians', 'Chennai Super Kings', 'Royal Challengers Bangalore',
        'Kolkata Knight Riders', 'Rajasthan Royals', 'Delhi Capitals',
        'Sunrisers Hyderabad'
    ],
    'home_city': [
        'Mumbai', 'Chennai', 'Bengaluru', 'Kolkata', 'Jaipur', 'Delhi', 'Hyderabad'
    ]
})

ipl_df = pd.merge(
    ipl_df,
    team_cities.rename(columns={'team': 'team1', 'home_city': 'team1_home_city'}),
    on='team1',
    how='left'
)

ipl_df = pd.merge(
    ipl_df,
    team_cities.rename(columns={'team': 'team2', 'home_city': 'team2_home_city'}),
    on='team2',
    how='left'
)

print(ipl_df[['team1', 'team1_home_city', 'team2', 'team2_home_city']])
```

```python
wins_pivot = pd.pivot_table(
    ipl_df,
    index='winner',
    columns='season',
    values='id',
    aggfunc='count',
    fill_value=0
)

print('Wins by team per season:')
print(wins_pivot)

top_3 = wins_pivot.stack().reset_index(name='wins').sort_values('wins', ascending=False).head(3)

print('\nTop 3 team-season combinations:')
print(top_3)
```
