```python
import pandas as pd

date_series = pd.Series([
    '12-03-2026',
    '25-05-2026',
    '07-07-2026',
    '18-08-2026',
    '30-09-2026'
])

date_series = pd.to_datetime(date_series, format='%d-%m-%Y')

print(date_series)
```

```python
movies = pd.DataFrame({
    'movie': ['Movie A', 'Movie B', 'Movie C', 'Movie D'],
    'release_date': ['15-01-2024', '20-05-2023', '10-08-2025', '25-12-2022']
})

movies['release_date'] = pd.to_datetime(movies['release_date'], format='%d-%m-%Y')
movies['year'] = movies['release_date'].dt.year
movies['month'] = movies['release_date'].dt.month
movies['day'] = movies['release_date'].dt.day

print(movies)
```

```python
today = pd.Timestamp.today().normalize()
next_7_days = pd.date_range(start=today, periods=7)

print(next_7_days.tolist())
```

```python
zomato_gold = pd.DataFrame({
    'user_id': [101, 102, 103, 104],
    'start_date': ['01-01-2026', '15-03-2026', '10-06-2026', '20-08-2026']
})

zomato_gold['start_date'] = pd.to_datetime(zomato_gold['start_date'], format='%d-%m-%Y')
today = pd.Timestamp.today().normalize()
zomato_gold['tenure_days'] = (today - zomato_gold['start_date']).dt.days

print(zomato_gold)
```
