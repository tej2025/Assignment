```python
import pandas as pd

movies_df = pd.read_csv('movies.csv')

selected_movies = movies_df.loc[:, ['title', 'rating']]

print(selected_movies)
```

```python
flipkart_df = pd.read_csv('flipkart_products_new.csv')

first_10_products = flipkart_df.iloc[:10, :]

print(first_10_products)
```

```python
zomato_df = pd.read_csv('zomato_food_orders.csv')

high_value_orders = zomato_df[zomato_df['order_amount'] > 500]

print(high_value_orders)
```

```python
spotify_df = pd.read_csv('spotify_songs.csv')

selected_songs = spotify_df.query(
    'streams > 1000000 and duration < 180'
)

print(selected_songs)
```

```python
ipl_df = pd.read_csv('ipl_matches_filter.csv')

selected_matches = ipl_df[
    (ipl_df['total_runs'] >= 180) &
    (ipl_df['total_runs'] <= 220)
]

print(selected_matches)
```
