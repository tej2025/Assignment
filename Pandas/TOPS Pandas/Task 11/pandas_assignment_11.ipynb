```python
import pandas as pd

january_orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104],
    'customer': ['Amit', 'Priya', 'Rahul', 'Neha'],
    'amount': [450, 300, 650, 500]
})

february_orders = pd.DataFrame({
    'order_id': [105, 106, 107, 108],
    'customer': ['Karan', 'Pooja', 'Vivek', 'Anjali'],
    'amount': [700, 350, 550, 400]
})

all_orders = pd.concat([january_orders, february_orders], ignore_index=True)

print(all_orders)
```

```python
product_details = pd.DataFrame({
    'product_id': [101, 102, 103, 104, 105],
    'product': ['Laptop', 'Headphones', 'Smartphone', 'Keyboard', 'Mouse'],
    'price': [55000, 2500, 30000, 1800, 900]
})

product_ratings = pd.DataFrame({
    'rating': [4.5, 4.2, 4.6, 4.1, 4.4]
})

products_with_ratings = pd.concat([product_details, product_ratings], axis=1)

print(products_with_ratings)
```

```python
spotify_streams = pd.DataFrame({
    'user': ['Tej'] * 7 + ['Rahul'] * 7 + ['Priya'] * 7,
    'day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'] * 3,
    'streams': [120, 150, 100, 180, 200, 250, 220,
                90, 110, 130, 120, 160, 190, 170,
                200, 180, 220, 210, 250, 300, 280]
})

spotify_pivot = spotify_streams.pivot(index='user', columns='day', values='streams')

print(spotify_pivot)
```

```python
zomato_orders = pd.DataFrame({
    'restaurant': ['Food Hub', 'Food Hub', 'Food Hub', 'Pizza Point', 'Pizza Point', 'Pizza Point',
                   'Spice House', 'Spice House', 'Spice House'],
    'month': ['Jan', 'Feb', 'Mar', 'Jan', 'Feb', 'Mar', 'Jan', 'Feb', 'Mar'],
    'total_amount': [400, 500, 450, 600, 550, 700, 350, 450, 500]
})

average_orders = zomato_orders.pivot_table(
    index='restaurant',
    columns='month',
    values='total_amount',
    aggfunc='mean'
)

print(average_orders)
```

```python
jan_sales = pd.read_csv('sales_jan.csv')
feb_sales = pd.read_csv('sales_feb.csv')
mar_sales = pd.read_csv('sales_mar.csv')

all_sales = pd.concat([jan_sales, feb_sales, mar_sales], ignore_index=True)

total_sales = all_sales.groupby('product')['sales'].sum().reset_index()

print('Combined monthly sales:')
print(all_sales)

print('\nTotal sales per product:')
print(total_sales)
```
