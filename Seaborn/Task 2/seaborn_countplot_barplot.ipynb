```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

swiggy_orders = pd.DataFrame({
    'Payment_Method': ['UPI', 'Card', 'Cash', 'UPI', 'Card', 'UPI', 'Cash', 'Card', 'UPI', 'Cash']
})

sns.countplot(x='Payment_Method', data=swiggy_orders)
plt.xlabel('Payment Method')
plt.ylabel('Number of Orders')
plt.title('Number of Orders by Payment Method')
plt.show()
```

```python
zomato_orders = pd.DataFrame({
    'Food_Category': ['Pizza', 'Burger', 'Biryani', 'South Indian', 'Pizza', 'Burger', 'Biryani', 'South Indian'],
    'Delivery_Time': [30, 25, 40, 35, 35, 30, 45, 30]
})

sns.barplot(x='Food_Category', y='Delivery_Time', data=zomato_orders, estimator='mean')
plt.xlabel('Food Category')
plt.ylabel('Average Delivery Time (Minutes)')
plt.title('Average Delivery Time by Food Category')
plt.show()
```

```python
sns.barplot(x='Food_Category', y='Delivery_Time', data=zomato_orders, estimator='sum')
plt.xlabel('Food Category')
plt.ylabel('Total Delivery Time (Minutes)')
plt.title('Total Orders by Food Category')
plt.show()
```

```python
bookmyshow_movies = pd.DataFrame({
    'Genre': ['Action', 'Comedy', 'Drama', 'Action', 'Romance', 'Comedy', 'Action', 'Drama', 'Comedy', 'Romance']
})

sns.countplot(x='Genre', data=bookmyshow_movies)
plt.xlabel('Movie Genre')
plt.ylabel('Number of Movies')
plt.title('Number of Movies by Genre')
plt.show()
```
