```python
import seaborn as sns
import matplotlib.pyplot as plt

delivery_times = [28, 32, 35, 30, 25, 40, 38, 29, 34, 31,
                  27, 36, 33, 42, 45, 39, 30, 28, 35, 37,
                  26, 32, 34, 41, 43, 29, 31, 36, 38, 33]

sns.histplot(delivery_times)
plt.xlabel("Delivery Time (Minutes)")
plt.ylabel("Number of Orders")
plt.title("Distribution of Swiggy Delivery Times")
plt.show()
```

```python
step_counts = [6500, 7200, 8100, 5900, 7600, 8500, 9000, 6800, 7400, 7900,
               8300, 6200, 7100, 8800, 9500, 6700, 7300, 8200, 7600, 6900,
               7800, 8600, 9200, 6400, 7500, 8000, 8700, 7000, 9300, 7200]

sns.kdeplot(step_counts)
plt.xlabel("Daily Step Count")
plt.ylabel("Probability Density")
plt.title("Distribution of Daily Step Counts")
plt.show()
```

```python
monthly_spending = [450, 620, 850, 1200, 980, 1500, 750, 1100, 1350, 900,
                    680, 1250, 1450, 800, 1050, 1600, 720, 950, 1300, 1150,
                    600, 1400, 1000, 780, 1550, 890, 1180, 1350, 920, 1280]

sns.histplot(monthly_spending, kde=True)
plt.xlabel("Monthly Spending (Rupees)")
plt.ylabel("Number of Orders")
plt.title("Distribution of Monthly Zomato Spending")
plt.show()
```

```python
ipl_scores = [85, 120, 145, 98, 175, 132, 110, 156, 88, 142,
             190, 105, 125, 160, 78, 135, 150, 115, 180, 95,
             130, 155, 100, 170, 90, 140, 165, 108, 125, 148]

sns.histplot(ipl_scores, bins=5)
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.title("IPL Run Distribution - 5 Bins")
plt.show()
```

```python
sns.histplot(ipl_scores, bins=10)
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.title("IPL Run Distribution - 10 Bins")
plt.show()
```

```python
sns.histplot(ipl_scores, bins=20)
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.title("IPL Run Distribution - 20 Bins")
plt.show()
```

### Observation

With 5 bins, the distribution is simple and gives a broad view of the scores.
With 10 bins, more details become visible, while 20 bins shows smaller changes and patterns in the data.
