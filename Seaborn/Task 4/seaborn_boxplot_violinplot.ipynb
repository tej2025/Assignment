```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

flipkart_data = pd.read_csv("flipkart_products.csv")

sns.boxplot(y='Price', data=flipkart_data)
plt.xlabel("Product Price")
plt.ylabel("Price (Rupees)")
plt.title("Flipkart Product Price Distribution")
plt.show()
```

```python
zomato_orders = pd.DataFrame({
    'Restaurant_Category': ['Fast Food', 'Fast Food', 'Fast Food', 'Fast Food', 'Fast Food',
                           'Desserts', 'Desserts', 'Desserts', 'Desserts', 'Desserts'],
    'Order_Amount': [180, 250, 320, 220, 290, 120, 180, 150, 210, 170]
})

sns.violinplot(x='Restaurant_Category', y='Order_Amount', data=zomato_orders)
plt.xlabel("Restaurant Category")
plt.ylabel("Order Amount (Rupees)")
plt.title("Zomato Order Amount by Restaurant Category")
plt.show()
```

```python
ticket_prices = [500, 750, 800, 1000, 1200, 1500, 900, 1100, 1300, 950,
                850, 1250, 1400, 1600, 1800, 2000, 2200, 5000]

sns.boxplot(y=ticket_prices)
plt.xlabel("IPL Match Tickets")
plt.ylabel("Ticket Price (Rupees)")
plt.title("IPL Match Ticket Price Distribution")
plt.show()
```

```python
# The point outside the whiskers represents an outlier.
print("The ticket price of Rs. 5000 is an outlier.")
```
