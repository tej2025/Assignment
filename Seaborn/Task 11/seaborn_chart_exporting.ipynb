```python
import seaborn as sns
import matplotlib.pyplot as plt

categories = ['Pizza', 'Burger', 'Biryani', 'Momos', 'Ice Cream']
orders = [85, 72, 95, 60, 48]

sns.barplot(x=categories, y=orders)
plt.title('Number of Orders by Food Category')
plt.xlabel('Food Category')
plt.ylabel('Number of Orders')

plt.savefig('food_orders.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
steps = [5200, 6100, 5800, 7200, 6800, 8100, 7600]

sns.lineplot(x=days, y=steps, marker='o')
plt.title('Daily Step Counts for One Week')
plt.xlabel('Day')
plt.ylabel('Steps')

plt.savefig('weekly_steps.svg', format='svg', bbox_inches='tight')
plt.show()
```

```python
sns.set_style('darkgrid')
sns.set_palette('Set2')

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [120, 150, 135, 180, 210]

sns.barplot(x=months, y=sales)
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.savefig('monthly_sales_1200dpi.png', dpi=1200, bbox_inches='tight')
plt.show()
```

```python
def apply_presentation_theme():
    sns.set(
        style='whitegrid',
        context='talk',
        palette='deep',
        font_scale=1.0
    )

apply_presentation_theme()

apps = ['WhatsApp', 'YouTube', 'Instagram', 'Chrome', 'Spotify']
usage = [30, 25, 20, 15, 10]

plt.pie(usage, labels=apps, autopct='%1.1f%%', startangle=90)
plt.title('Top 5 Most-Used Apps')
plt.savefig('top_5_apps.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd

data = pd.DataFrame({
    'Math': [85, 90, 78, 92],
    'Science': [88, 85, 80, 95],
    'English': [82, 87, 85, 90],
    'Computer': [95, 92, 88, 98]
})

sns.set_style('white')
sns.heatmap(data.corr(), annot=True, cmap='coolwarm')

plt.title('Subject Correlation Heatmap')
plt.savefig('subject_heatmap.svg', format='svg', dpi=1200, bbox_inches='tight')
plt.show()
```
