```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(10)

food_data = pd.DataFrame({
    'delivery_time': np.random.randint(20, 60, 20),
    'order_value': np.random.randint(150, 1000, 20),
    'tip_amount': np.random.randint(10, 100, 20),
    'rating': np.round(np.random.uniform(3.0, 5.0, 20), 1),
    'delivery_distance': np.round(np.random.uniform(1, 15, 20), 1)
})

correlation_matrix = food_data.corr()
print(correlation_matrix)
```

```python
sns.heatmap(correlation_matrix, annot=True)
plt.title("Correlation Matrix of Food Delivery Data")
plt.show()
```

```python
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title("Food Delivery Correlation Heatmap")
plt.show()
```

```python
corr_values = correlation_matrix.copy()
np.fill_diagonal(corr_values.values, np.nan)

strongest_negative = corr_values.stack().idxmin()
print("Strongest negative correlation is between", strongest_negative[0], "and", strongest_negative[1])
```

The two features with the strongest negative correlation are identified from the correlation matrix above.
A negative correlation means that when one feature increases, the other tends to decrease.

```python
flipkart_data = pd.read_csv("flipkart_mobile_phones.csv")

numeric_data = flipkart_data[['price', 'rating', 'RAM', 'battery_capacity']]
flipkart_corr = numeric_data.corr()

print(flipkart_corr)
```
