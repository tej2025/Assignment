```python
import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
songs = [45, 60, 52, 70, 85, 110, 95]

sns.set(context='poster')

sns.barplot(x=days, y=songs)
plt.title('Spotify Songs Streamed Per Day')
plt.xlabel('Day')
plt.ylabel('Number of Songs')
plt.show()
```

```python
cuisine = ['Indian', 'Chinese', 'Italian', 'Indian', 'Chinese',
           'Indian', 'Italian', 'Indian', 'Chinese', 'Italian',
           'Indian', 'Chinese', 'Indian', 'Italian', 'Chinese']

sns.set_palette('colorblind')

sns.countplot(x=cuisine, palette='colorblind')
plt.title('Food Orders by Cuisine Type')
plt.xlabel('Cuisine Type')
plt.ylabel('Number of Orders')
plt.show()
```

```python
days = list(range(1, 15))
steps = [5200, 6100, 5800, 7200, 6800, 7500, 8100,
         6400, 7000, 7600, 8300, 7900, 8800, 9200]

sns.set(context='notebook', palette='muted')

sns.lineplot(x=days, y=steps)
plt.title('Daily Step Counts - 2 Weeks')
plt.xlabel('Day')
plt.ylabel('Steps')
plt.show()
```

```python
products = ['Mobile', 'Laptop', 'Tablet', 'Watch', 'Headphones']
sales = [120, 95, 80, 65, 110]

sns.set(context='talk', palette='deep')

sns.barplot(x=products, y=sales)
plt.title('Sales - Deep Palette and Talk Context')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.show()

sns.set(context='poster', palette='colorblind')

sns.barplot(x=products, y=sales)
plt.title('Sales - Colorblind Palette and Poster Context')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.show()
```

```python
sns.set(
    context='talk',
    style='whitegrid',
    palette='deep',
    font_scale=1.1
)

categories = ['Mobiles', 'Laptops', 'Fashion', 'Home', 'Electronics']
sales = [450000, 380000, 320000, 280000, 410000]

sns.barplot(x=categories, y=sales)
plt.title('Flipkart Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Sales (Rs.)')
plt.show()
```
