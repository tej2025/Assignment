```python
import seaborn as sns
import matplotlib.pyplot as plt

print("Seaborn version:", sns.__version__)
print("Matplotlib version:", plt.matplotlib.__version__)
```

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.lineplot(x=days, y=temps)

plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Weekly Temperature Trend")
plt.show()
```

```python
# Axes-level function
sns.lineplot(x=days, y=temps)

plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Weekly Temperature Trend - Axes-level")
plt.show()
```

```python
# Figure-level function
sns.relplot(x=days, y=temps, kind="line")

plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Weekly Temperature Trend - Figure-level")
plt.show()
```

```python
# Difference:
# Axes-level functions work on a specific matplotlib Axes,
# while figure-level functions create and manage the entire figure.
```

```python
# Darkgrid theme
sns.set_theme(style="darkgrid")

sns.lineplot(x=days, y=temps)
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Weekly Temperature Trend - Darkgrid")
plt.savefig("temperature_darkgrid.png")
plt.show()
```

```python
# Whitegrid theme
sns.set_theme(style="whitegrid")

sns.lineplot(x=days, y=temps)
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Weekly Temperature Trend - Whitegrid")
plt.savefig("temperature_whitegrid.png")
plt.show()
```

```python
# Ticks theme
sns.set_theme(style="ticks")

sns.lineplot(x=days, y=temps)
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Weekly Temperature Trend - Ticks")
plt.savefig("temperature_ticks.png")
plt.show()
```

```python
# Comparison:
# darkgrid: Dark background with grid lines.
# whitegrid: White background with grid lines.
# ticks: Clean background with axis tick marks and no grid lines.
```
