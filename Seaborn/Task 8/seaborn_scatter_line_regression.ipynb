```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

instagram_data = pd.DataFrame({
    'followers': [500, 1200, 2500, 3400, 5000, 6500, 7200, 8500, 10000, 12000,
                  15000, 18000, 21000, 25000, 30000, 35000, 42000, 50000, 65000, 80000],
    'posts': [45, 80, 120, 150, 210, 260, 300, 350, 420, 480,
              550, 620, 700, 780, 850, 920, 1050, 1200, 1350, 1500]
})

sns.scatterplot(x='followers', y='posts', data=instagram_data)
plt.xlabel("Number of Followers")
plt.ylabel("Number of Posts")
plt.title("Followers vs Number of Posts")
plt.show()
```

```python
days = list(range(1, 31))
order_counts = [120, 135, 128, 150, 142, 160, 155, 170, 145, 138,
                152, 165, 158, 175, 180, 168, 172, 185, 190, 178,
                160, 155, 170, 182, 195, 188, 200, 192, 205, 198]

sns.lineplot(x=days, y=order_counts)
plt.xlabel("Day")
plt.ylabel("Number of Food Orders")
plt.title("Daily Zomato Food Orders")
plt.show()
```

```python
movies = pd.DataFrame({
    'movie': ['Movie 1', 'Movie 2', 'Movie 3', 'Movie 4', 'Movie 5',
              'Movie 6', 'Movie 7', 'Movie 8', 'Movie 9', 'Movie 10',
              'Movie 11', 'Movie 12', 'Movie 13', 'Movie 14', 'Movie 15'],
    'rating': [4.2, 3.8, 4.5, 3.9, 4.1, 4.6, 3.7, 4.3, 4.0, 4.7,
               3.6, 4.4, 4.1, 3.9, 4.5],
    'reviews': [1200, 850, 2100, 950, 1400, 2500, 700, 1800, 1100, 2800,
                650, 2200, 1300, 900, 2000],
    'genre': ['Action', 'Comedy', 'Drama', 'Action', 'Comedy',
              'Drama', 'Action', 'Comedy', 'Drama', 'Action',
              'Comedy', 'Drama', 'Action', 'Comedy', 'Drama']
})

sns.relplot(x='rating', y='reviews', hue='genre', data=movies)
plt.xlabel("Movie Rating")
plt.ylabel("Number of Reviews")
plt.show()
```

```python
students = pd.DataFrame({
    'youtube_hours': [1, 2, 1.5, 3, 2.5, 4, 3.5, 5, 4.5, 6,
                      5.5, 7, 6.5, 8, 7.5, 2, 3, 4, 5, 6,
                      7, 8.5, 9, 2.5, 4.5],
    'exam_score': [92, 88, 90, 84, 86, 80, 82, 76, 79, 72,
                   75, 68, 70, 64, 66, 87, 83, 81, 77, 73,
                   69, 61, 58, 85, 78]
})

sns.regplot(x='youtube_hours', y='exam_score', data=students)
plt.xlabel("Hours Spent on YouTube")
plt.ylabel("Exam Score")
plt.title("YouTube Hours vs Exam Scores")
plt.show()
```

### Interpretation

The regression line shows a negative trend. As the number of hours spent on YouTube increases, exam scores generally decrease.
