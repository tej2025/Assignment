# Bank Marketing Dataset Analysis

## 1. Load the Kaggle Bank Marketing Dataset

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

bank_df = pd.read_csv('bank-full.csv', sep=';')

print(bank_df.head(10))
print(bank_df.info())
```

## 2. Term Deposit Subscription Countplot

```python
sns.countplot(data=bank_df, x='y')

plt.title('Term Deposit Subscription')
plt.xlabel('Subscribed to Term Deposit')
plt.ylabel('Number of Customers')
plt.show()
```

## 3. Age and Term Deposit Subscription

```python
sns.boxplot(data=bank_df, x='y', y='age')

plt.title('Age vs Term Deposit Subscription')
plt.xlabel('Subscribed to Term Deposit')
plt.ylabel('Age')
plt.show()
```

## 4. Percentage of Customers Subscribed by Education Level

```python
education_percentage = (
    bank_df.groupby('education')['y']
    .apply(lambda x: (x == 'yes').mean() * 100)
    .reset_index(name='subscription_percentage')
)

sns.barplot(
    data=education_percentage,
    x='education',
    y='subscription_percentage'
)

plt.title('Term Deposit Subscription Percentage by Education')
plt.xlabel('Education Level')
plt.ylabel('Subscription Percentage (%)')
plt.xticks(rotation=45)
plt.show()
```

## 5. Additional Visualizations for a Fintech Savings Product

```python
job_percentage = (
    bank_df.groupby('job')['y']
    .apply(lambda x: (x == 'yes').mean() * 100)
    .reset_index(name='subscription_percentage')
    .sort_values('subscription_percentage', ascending=False)
)

sns.barplot(
    data=job_percentage,
    x='job',
    y='subscription_percentage'
)

plt.title('Term Deposit Subscription Percentage by Job')
plt.xlabel('Job Type')
plt.ylabel('Subscription Percentage (%)')
plt.xticks(rotation=45)
plt.show()
```
