```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

instagram_data = pd.DataFrame({
    'likes': [120, 250, 180, 420, 350, 500, 290, 650, 400, 220, 550, 310],
    'comments': [12, 25, 18, 40, 32, 48, 27, 60, 38, 20, 52, 29]
})

sns.regplot(x='likes', y='comments', data=instagram_data)
plt.xlabel("Number of Likes")
plt.ylabel("Number of Comments")
plt.title("Likes vs Comments on Instagram Posts")
plt.show()
```

```python
zomato_data = pd.DataFrame({
    'price': [180, 250, 320, 450, 220, 380, 500, 290, 350, 420, 200, 310, 470, 260, 390],
    'delivery_time': [25, 28, 32, 40, 27, 35, 45, 30, 34, 38, 24, 31, 42, 29, 36]
})

sns.lmplot(x='price', y='delivery_time', data=zomato_data)
plt.xlabel("Order Price (Rupees)")
plt.ylabel("Delivery Time (Minutes)")
plt.show()
```

```python
spotify_data = pd.DataFrame({
    'songs': [5, 8, 10, 12, 15, 18, 20, 22, 25, 28, 30, 35, 40, 45, 50],
    'duration': [18, 30, 38, 45, 55, 68, 72, 85, 95, 105, 115, 135, 150, 170, 190]
})

sns.regplot(x='songs', y='duration', data=spotify_data, lowess=True)
plt.xlabel("Number of Songs")
plt.ylabel("Total Playlist Duration (Minutes)")
plt.title("Songs vs Playlist Duration")
plt.show()
```

```python
flipkart_data = pd.DataFrame({
    'price': [999, 1499, 1999, 2499, 2999, 3499, 3999, 4499, 4999, 5499],
    'reviews': [850, 720, 680, 590, 510, 450, 390, 330, 280, 240]
})

sns.lmplot(x='price', y='reviews', data=flipkart_data, scatter_kws={'color': 'green'})
plt.xlabel("Product Price (Rupees)")
plt.ylabel("Number of Reviews")
plt.title("Flipkart Product Price vs Number of Reviews")
plt.show()
```
